## 1. Los 5 perfiles principales

Cada perfil tiene una pregunta central que lo define:

```python
perfiles = {
    "Data Scientist": "¿Qué nos dicen los datos?",
    "ML Engineer": "¿Cómo entreno y sirvo mi modelo de forma fiable?",
    "Prompt Engineer": "¿Qué instrucciones hacen que el modelo se comporte como quiero?",
    "LLM Developer": "¿Cómo integro un LLM en esta feature?",
    "AI Engineer": "¿Cómo construyo un SISTEMA de IA confiable de punta a punta?"
}

for rol, pregunta in perfiles.items():
    print(f"✓ {rol:20} {pregunta}")
```

## 2. Ejemplo: clasificador simple de roles

Vamos a construir un clasificador basado en palabras clave.

```python
import re

palabras_clave = {
    "ML Engineer": ["entrenar", "mlops", "gpu", "fine-tuning"],
    "AI Engineer": ["arquitectura", "producción", "observabilidad", "rag"],
    "Prompt Engineer": ["prompt", "few-shot", "instrucciones"],
}

def clasificar_simple(tarea: str) -> str:
    """Clasifica una tarea contando palabras clave."""
    tarea = tarea.lower()
    puntajes = {}
    
    for rol, keywords in palabras_clave.items():
        puntajes[rol] = sum(1 for kw in keywords if kw in tarea)
    
    return max(puntajes, key=puntajes.get) if any(puntajes.values()) else "AI Engineer"

# Prueba
tareas_ejemplo = [
    "Entrenar un modelo con GPU y hacer fine-tuning",
    "Mejorar el prompt con ejemplos few-shot",
    "Diseñar arquitectura RAG con observabilidad"
]

for tarea in tareas_ejemplo:
    print(f"{tarea[:40]:40} → {clasificar_simple(tarea)}")
```

## 3. Lección importante

**No siempre necesitas un LLM.** Este clasificador usa reglas simples:
- Es más rápido (microsegundos vs milisegundos)
- Es más barato ($0 vs centavos por llamada)
- Es determinista (siempre da la misma respuesta)
- Es fácil de debuggear

Un buen AI Engineer sabe cuándo usar cada herramienta.
